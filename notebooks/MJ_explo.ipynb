```python
d={'code_insee': 'string',
   'nom_standard': 'string',
   'dep_code': 'string',
   'dep_nom': 'string',
   'superficie_km2': 'float32',
   'altitude_moyenne': 'float32'}
df=pd.read_csv('../data/communes-france-2025.csv.gz', compression='gzip',dtype=d,usecols=d)
```

```python
total_area_per_department=(
df.groupby(['dep_code','dep_nom'])
    .agg(departement_area_km2=('superficie_km2','sum'))
    .reset_index()
)
```

```python
df = df.merge(total_area_per_department, on=['dep_code', 'dep_nom'], how='left')
```

```python
df['area_share']=df['superficie_km2'] / df['departement_area_km2']
```

```python
df['mean_altitude_weighted']=df['altitude_moyenne']*df['area_share']
```

```python
df.groupby(['dep_code', 'dep_nom']).agg(total_superficie=('mean_altitude_weighted', 'sum')).reset_index()
```

```python
df.to_csv("result.csv")
```
